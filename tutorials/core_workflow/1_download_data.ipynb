# **TUTORIAL** 1: Set the workspace and download source data

The first step for using pyToCE is to create a workspace. This tutorial uses
`set_workspace(...)` to select that directory and install the packaged input
files, then `download_data(...)` to download public sources and register the
licensed ecoinvent archives supplied by the user.

## Before starting...

Python 3.11 or 3.12 is required.

On macOS with Apple Silicon, install the LCA solver build requirements using
[Homebrew](https://brew.sh/):

```console
brew install suite-sparse swig pkg-config
```

Install the package from PyPI:

```console
python -m pip install pytoce
```

Have the two licensed ecoinvent 3.12 archives available locally. Their required
filenames are given below.


````python
from pathlib import Path

from pytoce import download_data, set_workspace
````

## Select the workspace

Edit the path below and use the same workspace throughout the tutorials.
Every tutorial notebook repeats `set_workspace(...)`, as all function calls
assume it has already run in the current Python session.

The workspace contains three folders: `data_raw/` for source files and user
inputs, `data_processed/` for derived inputs reused by later calculations, and
`results/` for MFA, LCA, ASR, and optimization results. The report gives the
workspace paths, including those for scenario templates, allocated carrying
capacity inputs, and methodological notes.


````python
workspace = Path.home() / "Documents" / "pytoce_workspace"
workspace_report = set_workspace(workspace)
workspace_report.as_dict()
````

## Add the licensed ecoinvent files

An ecoinvent license and local copies of the following archives are required.
Place them in `data_raw/ecoinvent/` before calling `download_data(...)`:

- `ecoinvent_3.12_cutoff_ecoSpold02.7z`
- `ecoinvent 3.12_LCIA_implementation.7z`

Keep their original filenames and archive contents. The next cell shows the
directory in the selected workspace.


````python
workspace_report.data_raw / "ecoinvent"
````

## Download and check the sources

The function downloads and verifies the following public sources:

| Source | Version | Used for |
| --- | --- | --- |
| [REMIND-MFA](https://github.com/pik-piam/remind-mfa) | 0.2.0 | Dynamic MFA |
| [REMIND-MFA input data](https://doi.org/10.5281/zenodo.18019915) | 1.0.0 | Historical data and default parameter trajectories |
| [PB-LCIA](https://github.com/Ike-EDB/PB-LCIA/tree/ei_312) | 0.9.2 | PB-LCIA characterization factors for ecoinvent 3.12 |

It also verifies the licensed archives and writes
`data_processed/ecoinvent/material_recipe.json`. This recipe contains monomer
requirements, manufacturing coefficients, recovery yields, and carbon fractions
derived from selected EcoSpold records. The MFA and background preparation reuse
these coefficients.

The first call requires network access. Later calls reuse completed sources;
`refresh=True` downloads the cataloged public sources again. Source manifests
record the versions, revisions, licenses, and file hashes. Keep them with the
source files when reusing the workspace.


````python
download_report = download_data()
download_report.as_dict()
````
