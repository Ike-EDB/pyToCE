# **TUTORIAL** 2: Prepare the prospective backgrounds

Prospective background inventories represent changes in the supply chains used
to quantify environmental burdens over time. `prepare_backgrounds(...)` uses
[premise](https://github.com/polca/premise) to transform ecoinvent 3.12 inventories
using REMIND SSP2 projections. These inventories are used by the time-explicit
prospective LCA in Tutorial 4.

## Premise data

A premise decryption key is required to prepare new backgrounds from the
encrypted integrated assessment model (IAM) scenario files. Request the key
from the premise developers as described in the
[premise documentation](https://premise.readthedocs.io/en/latest/getting_started/preparation.html).

Store the key in the `PREMISE_KEY` environment variable before starting Jupyter.
The call below passes it through `premise_key`. Later calls reuse completed
backgrounds.


## Prepare the four backgrounds

pyToCE uses the four REMIND SSP2 background scenarios available in premise 2.4.6,
covering climate policy pathways ranging from below 1.5 °C to above
3 °C of warming by 2100:

- `remind_ssp2_ndc`
- `remind_ssp2_npi`
- `remind_ssp2_pkbudg1000`
- `remind_ssp2_pkbudg650`

The warming range follows indicative classifications in the premise scenario
catalog. Each background is prepared separately, and the report gives its status,
directory, and manifest. A later call reuses each completed background.


```python
background_report = prepare_backgrounds(
    premise_key=os.environ.get("PREMISE_KEY"),
)
background_report.as_dict()
```

To select one background or a sequence, pass its scenario name or names through
`background_scenario`, for example `background_scenario="remind_ssp2_npi"`.

Ecoinvent 3.12 is used as the general background system through 2025. For
2026--2029, linear interpolation provides a gradual transition between ecoinvent
in 2025 and the corresponding premise database in 2030. From 2030 onward,
intermediate years are represented by interpolation between successive premise
databases. Specific foreground technologies required before 2030 use inventories
from premise 2020 and 2025.
